```python
import pandas as pd
import numpy as np

from pathlib import Path
```

```python
def convert_df_ravenpro(df: pd.DataFrame):
    """
    Converts a dataframe to the format used by RavenPro
    """

    ravenpro_df = df.copy()

    ravenpro_df.rename(columns={
        "start_time": "Begin Time (s)",
        "end_time": "End Time (s)",
        "low_freq": "Low Freq (Hz)",
        "high_freq": "High Freq (Hz)",
        "event": "Annotation",
    }, inplace=True)

    ravenpro_df["Selection"] = np.arange(0, df.shape[0]).astype('int') + 1
    ravenpro_df["View"] = "Waveform 1"
    ravenpro_df["Channel"] = "1"

    return ravenpro_df
```

```python
bd2_df = pd.read_csv(f'{Path.home()}/duty-cycle-investigation/data/raw/Carp/bd2__Carp_20220717_013000.csv')
extension = ".txt"
sep = "\t"

convert_df_ravenpro(bd2_df).to_csv(f'{Path.home()}/Downloads/bd2__Carp_20220717_013000.txt', sep=sep, index=False)
```

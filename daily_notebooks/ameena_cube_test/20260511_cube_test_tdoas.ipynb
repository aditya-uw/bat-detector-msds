The following cell:

- plots the microphone positions again

- creates the intervals of sound based on the call times
- cross correlates the intervals
- localizes
- prints the number of calls in the set
- for each call:
    - prints the call index number within the set
    - prints the call time
    - prints the rows in order of largest amplitude during cross correlation
    - prints the estimated location based on each simulated call
    - prints error of each point (linear distance from estimated point to actual call)
- prints the mean error
- prints the actual source coordinates
- prints the mean estimation ("mean calculated")

The way that the process works/things you can customize:
- uncomment the set of calls that you want to use as the set of call times

- for every time in that set, it will find the difference between the time between the first time in the CSV for each mic (ex: if the call time is 1:25:00 and the first time in the CSV for a mic is 1:20:00, the delta is 0:05:00 or five minutes)
- it will convert that time difference to seconds (five minutes --> 300 seconds)
- it will say that the start of the interval for cross correlation is that many seconds into the sound file (the interval starts at 300 seconds into the .WAV file)
- you can change the interval length, but since the test signals are 1 second long it is best to leave it at 2 seconds as is (continuing with the example, the interval will start at 300 seconds into the .WAV file and end at 302 seconds into the .WAV file)
- an NxM array of all the intervals is created where N is the number of microphones and M is the interval in samples (aka 2 seconds * fs)
- it then cross correlates/finds the lag between each interval and the interval of the reference mic in units of samples (aka tdoa * fs)
- it then creates a matrix and solution vector to solve for the coordinates
- it will sort the rows of the matrix in order of how large the lag amplitude was during the cross correlation, in order to prioritize using lags that had a larger "spike"
- it will use the five rows with the largest "spike" and their respective solution vector values and use that to solve for the source coordinates. Even though it is a 4x4 matrix, it is best to use the last 5 rows in case that the row generated by cross correlating the reference mic's interval with itself creates one of the largest spikes, otherwise there will be only three valid rows and one row of zeros which cannot be used
- it will plot the point, with the option to plot line segments in the x, y, then z  directions in order to help visualize where the point is relative to the reference mic. To do that, uncomment the three lines under "##to make it clear where the point is:"
- the error of each estimated point is calculated as the shortest linear distance from the estimation to the actual source
- the actual source is plotted in blue (this color can be changed by changing "theoretical_color")
- the estimated locations are plotted in yellow (this color can be changed by changing "localized color")
- the mean estimation, calculated as (mean(all x), mean(all y), mean(all z)) is plotted in red
- the mean error is calculated as the mean of all the errors (mean of all the distances)
- there is also the option to print spectrograms for each call, which is controlled by uncommenting the lines under "##print spectrograms:"

```python
scale_fac = (2**40 / 10**12)
```

```python
HARD_DRIVE_SIZES_TiB = {'ubna_data_01': 6.849,
                    'ubna_data_02': 15.018,
                    'ubna_data_03': 15.486, 
                    'ubna_data_04': 17.206,
                    'ubna_data_05': 6.826,
                    'ubna_data_06': 15.439,
                    'ubna_data_07': 13.988}
```

```python
HARD_DRIVE_SIZES_TB = dict((k, v * scale_fac) for k, v in HARD_DRIVE_SIZES_TiB.items())
HARD_DRIVE_SIZES_TB
```

```python
sum_all = 0
for k, v in HARD_DRIVE_SIZES_TB.items():
    print(f"{k}: {v:.3f} TB")
    sum_all += v
print(f"Total: {sum_all:.3f} TB")
```

```python
bd2_file_df = pd.read_csv('20250130__group_threshold_sweep_results/bd2__threshold0p00_chunksize2_20220826_070000_REDUCED_OVERLAPS.csv', index_col=0)
selected_bd2_file_df = bd2_file_df[(bd2_file_df['det_prob']>=0.35)&(bd2_file_df['SNR']>=3)]
selected_bd2_file_df
```

```python
def convert_bd2df_ravenpro(df: pd.DataFrame):
    """
    Converts a dataframe to the format used by RavenPro
    """

    ravenpro_df = df.copy()

    ravenpro_df.rename(columns={
        "start_time": "Begin Time (s)",
        "end_time": "End Time (s)",
        "low_freq": "Low Freq (Hz)",
        "high_freq": "High Freq (Hz)",
        "event": "Annotation",
        "freq_group":"Manually-Verified Phonic Group",
        'snr_nist_quick_dB':'SNR NIST Quick (dB)',
        'delta_time_s':'Delta Time (s)'
    }, inplace=True)

    ravenpro_df["Selection"] = np.arange(0, df.shape[0]).astype('int') + 1
    ravenpro_df["View"] = "Waveform 1"
    ravenpro_df["Channel"] = "1"

    return ravenpro_df
```

```python
selected_bd2_file_txt = convert_bd2df_ravenpro(selected_bd2_file_df)
selected_bd2_file_txt
```

```python
import numpy as np
import matplotlib.pyplot as plt

from torch import multiprocessing
from pathlib import Path
```

```python
times_taken = np.load('20241106__large_instance_60chunks_16x16_computation.npy')
```

```python
plt.figure(figsize=(8,6))
plt.plot(times_taken.T, marker='.')
plt.xlabel('Chunksize (chunks per processor)')
plt.ylabel('Time taken (s)')
line_labs = []
for p in range(16):
    line_labs+=[f'{int(p+1)} processes']
plt.legend(line_labs, ncols=2)
plt.xticks(np.arange(multiprocessing.cpu_count()), np.arange(multiprocessing.cpu_count())+1)
plt.grid(which='both')
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.title(f'60 segments fixed; num_threads=1')
plt.imshow(times_taken/60)
plt.ylabel('Num_processes (processors assigned)')
plt.xlabel('Chunksize (chunks per processor)')
plt.yticks(np.arange(multiprocessing.cpu_count())-0.5, np.arange(multiprocessing.cpu_count())+1)
plt.xticks(np.arange(multiprocessing.cpu_count())-0.5, np.arange(multiprocessing.cpu_count())+1)
plt.grid(which='both')
plt.colorbar(label='Time taken (min)')
plt.savefig(f'20241106__large_instance_60chunks_{multiprocessing.cpu_count()}x{multiprocessing.cpu_count()}_computation.png', bbox_inches='tight')
# plt.close()
plt.show()
```

```python
import pandas as pd
import dask.dataframe as dd
import datetime as dt

from pathlib import Path

import numpy as np
```

```python
ubna_data_01_df = pd.read_csv('../output_dir/ubna_data_01_collected_audio_records.csv', index_col=0)
dc_season1 = ubna_data_01_df[pd.to_datetime(ubna_data_01_df['Datetime UTC'])>=dt.datetime(2022,1,1)]
dc_season1.head(5)
```

```python
ubna_data_02_df = pd.read_csv('../output_dir/ubna_data_02_collected_audio_records.csv', index_col=0)
dc_season2 = ubna_data_02_df[pd.to_datetime(ubna_data_02_df['Datetime UTC'])<=dt.datetime(2022,10,18,0,0,0)]
dc_season2.head(5)
```

```python
dc_season = pd.concat([dc_season1, dc_season2])
dc_season["Datetime UTC"] = pd.DatetimeIndex(dc_season["Datetime UTC"])
dc_season.set_index("Datetime UTC", inplace=True)
dc_season
```

```python
def filter_df_with_location(ubna_data_df, cfg):
    site_name_cond = ubna_data_df["Site name"] == cfg['site']
    file_year_cond = ubna_data_df.index.year == (dt.datetime.strptime(cfg['year'], '%Y')).year
    minute_cond = np.logical_or((ubna_data_df.index).minute == 30, (ubna_data_df.index).minute == 0)
    datetime_cond = np.logical_and((ubna_data_df.index).second == 0, minute_cond)
    file_error_cond = np.logical_and((ubna_data_df["File duration"]!='File has no comment due to error!'), (ubna_data_df["File duration"]!='File has no Audiomoth-related comment'))
    all_errors_cond = np.logical_and((ubna_data_df["File duration"]!='Is empty!'), file_error_cond)

    filtered_location_df = ubna_data_df.loc[site_name_cond&datetime_cond&file_year_cond&all_errors_cond].sort_index()
    filtered_location_nightly_df = filtered_location_df.between_time(cfg['recording_start'], cfg['recording_end'], inclusive="left")

    return filtered_location_nightly_df
```

```python
hour_recorded = dict()
hours_recorded_sum = 0
sites = ['Carp Pond', 'Telephone Field', 'Foliage', 'Central Pond', 'Fallen Tree', 'Opposite Central Pond']
```

```python
site = sites[0]
cfg  = dict()
cfg["site"] = site
cfg["duration"] = 1795
cfg['year'] = '2022'
cfg['recording_start'] = '00:00'
cfg['recording_end'] = '16:00'

files_from_location = filter_df_with_location(dc_season, cfg)

ref_audio_files = sorted(list(files_from_location["File path"].apply(lambda x : Path(x)).values))
file_status_cond = files_from_location["File status"] == "Usable for detection"
file_duration_cond = np.isclose(files_from_location["File duration"].astype('float'), cfg['duration'])
good_location_df = files_from_location.loc[file_status_cond&file_duration_cond]
good_audio_files = sorted(list(good_location_df["File path"].apply(lambda x : Path(x)).values))
files_in_dc_study = good_location_df.loc[~good_location_df.index.duplicated(keep='last')]
seconds_recorded_in_dc_study = files_in_dc_study['File duration'].astype(np.float64)
minutes_recorded_in_dc_study = seconds_recorded_in_dc_study/60
hours_recorded_in_dc_study = minutes_recorded_in_dc_study/60
assert(len(sorted(list((Path.home() / f'duty-cycle-analysis/data/raw/{cfg["site"].split()[0]}').glob('bd2_*.csv')))) == good_location_df.loc[~good_location_df.index.duplicated(keep='last')].shape[0])
hour_recorded[cfg['site']] = hours_recorded_in_dc_study.sum()
hour_recorded[cfg['site']]
```

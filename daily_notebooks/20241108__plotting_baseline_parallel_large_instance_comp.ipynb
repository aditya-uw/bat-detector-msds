```python
import numpy as np
import matplotlib.pyplot as plt

from torch import multiprocessing
from pathlib import Path
```

```python
baseline_times_taken = np.load('20241107__large_160gb_baseline_16p1c1t_estimates.npy')
parallel_times_taken = np.load('20241107__large_160gb_parallel_16p1c1t_estimates.npy')
```

```python
plt.figure(figsize=(8,6))
plt.rcParams.update({'font.size':14})
plt.title(f'Large 160GB Disk 60GB RAM 16-core instance')
plt.bar(x=np.arange(3)+0.625, height=baseline_times_taken[::-1]/60, width=0.25)
plt.bar(x=np.arange(3)+0.375, height=parallel_times_taken[::-1]/60, width=0.25)
plt.xticks(ticks=np.arange(3)+0.5, labels=['Generating segments', 'Running BatDetect2', 'Overall pipeline'][::-1])
line_labs = ['Baseline (no-parallelization)', 'Parallelization (all available workers)']
plt.legend(line_labs, ncols=1, loc='upper right')
plt.yticks(np.arange(0, 30000, 3000)/60)
plt.xlabel('Major pipeline tasks')
plt.ylabel('Time taken (min)')
plt.grid(which='both')
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.rcParams.update({'font.size':14})
plt.bar(x=np.arange(3)+0.5, height=(baseline_times_taken[::-1]/60)/(parallel_times_taken[::-1]/60), width=0.25)
plt.xticks(ticks=np.arange(3)+0.5, labels=['Generating segments', 'Running BatDetect2', 'Overall pipeline'][::-1])
line_labs = ['160GB Disk 60GB RAM 16-core instance']
plt.legend(line_labs, ncols=1, loc='upper left')
plt.xlabel('Major pipeline tasks')
plt.ylabel('Speed improvement')
plt.grid(which='both')
plt.show()
```
